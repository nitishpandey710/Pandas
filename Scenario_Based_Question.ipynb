## 1: You are given two sentences as strings.
Create a function that returns a list of all words that are not in both sentences.
You can assume that the sentences have no punctuation marks, extra tabs, or spaces.

Note: The strings "Game" and "game" should be considered the same word. Because of this, make sure all the strings in the list you output are in lowercase.

Example:
Input:

string1 = "I want to eat fish today with my friends"
string2 = "All I want is to eat fish today with my friends"
Output:
def mismatched_words(string1,string2) -> ["all","is"].

```python
string1 = "I want to eat fish today with my friends" 
string2 = "All I want is to eat fish today with my friends" 

print(string1.lower().split())

print(string2.lower().split())

print(set(string1.lower().split()))
print(set(string2.lower().split()))
```

```python
def mismatched_words(string1,string2):
    words1=set(string1.lower().split())
    words2=set(string2.lower().split())
    mismatched=words1.symmetric_difference(words2)
    return list(mismatched)
```

```python
mismatched_words(string1,string2)
```

## 2: You are given a string that resembles the declaration of a list of dictionaries.

Without using the pandas package, write a function read_split_from_str to split the data into two lists, one for training and one for testing,
with a 70:30 split between the training set and the testing set.

Note: Use numpy for random number generation if you need to.
Example
Input:
list_of_dict_str = "[{'x': 0.0, 'y': 5.43}, {'x': 50.0, 'y': 102.78}, {'x': 100.0, 'y': 204.24}]"

```python
def read_split_from_str(list_of_dict_str):
    list_of_dict = eval(list_of_dict_str)
    split_point = len(list_of_dict) * 7 // 10
    return [list_of_dict[split_point:], list_of_dict[:split_point]]
```

```python
read_split_from_str(list_of_dict_str)
```

## 4: Write a function to select only the rows where the student’s favorite color is green or red and their grade is above 90.

```python
data = {
    'Name': ['Nitish', 'Mike', 'John', 'Peter', 'Mathews'],
    'FavoriteColor': ['red', 'blue', 'green', 'green', 'yellow'],
    'Grade': [95, 88, 92, 85, 91]
}

student_df = pd.DataFrame(data)

def grade_color(df):
    filtered_df=df[df['FavoriteColor'].isin(['red','green']) & (df['Grade']> 90)]
    return filtered_df
```

## 5: You are given a dataset of students and their scores in a subject. Write a function that
<br>Adds two new columns to the DataFrame:
<br>One showing the rank of each student based on their score (highest score = rank 1), allowing gaps in rank.
    <br>Another showing the dense rank, where there are no gaps in ranking.
    <br>Returns the DataFrame sorted by Score in descending order.

```python
data={
    'Student':['Nitish', 'Mike', 'John', 'Peter', 'Mathews','Pinto'],
    'Score':[95,80,95,70,60,80]
}

student_df=pd.DataFrame(data)
student_df
```

```python
student_df['Rank']=student_df['Score'].rank(ascending=False,method='min')
```

```python
student_df
```

```python
student_df['Dense_Rank']=student_df['Score'].rank(ascending=False,method='dense')
```

```python
student_df
```

```python
def ranking(df):
    student_df.sort_values(by='Score',ascending=False).reset_index(drop=True)
    student_df['Rank']=student_df['Score'].rank(ascending=False,method='min')
    student_df['Dense_rank']=student_df['Score'].rank(ascending=False,method='dense')
    student_df['lag']=student_df['Score'].shift(1)
    student_df['lead']=student_df['Score'].shift(-1)
    return student_df
```

```python
ranking(student_df)
```

## 10: Given a DataFrame with a Salary column, identify outliers using the IQR (Interquartile Range) method.

```python
data={
    'salary':[45000, 47000, 48000, 50000, 52000, 150000, 55000, 60000, 62000]
}
```

```python
df=pd.DataFrame(data)
```

```python
df
```

```python
#IQR Method
Q1=df['salary'].quantile(0.25)
Q3=df['salary'].quantile(0.75)

IQR=Q3 - Q1
```

```python
Q1
```

```python
Q3
```

```python
IQR
```

```python
lower_bound=Q1-1.5*IQR
Upper_bound=Q3+1.5*IQR
```

```python
Outliers=df[(df['salary'] < lower_bound) | (df['salary'] > Upper_bound)]
```

## 12: Fill missing Salary values using the average salary of the same Department.

```python
#Impute missing Salary with department-wise mean

import numpy as np
data = {
    'Name': ['Alice', 'Bob', 'Prince', 'David', 'Balu', 'Jackson', 'Grace'],
    'Age': [25, 30, np.nan, 40, 29, np.nan, np.nan],
    'Salary': [50000, 54000, 60000, np.nan, np.nan, np.nan, np.nan],
    'Department': ['HR', 'IT', 'HR', 'IT', 'Finance', 'Finance', 'HR']
}

df=pd.DataFrame(data)
```

```python
df
```

```python
df['Salary'] = df.groupby('Department')['Salary'].transform(lambda x: x.fillna(x.mean()))
```
